# 06 ? Intent classification

Compare a word+character TF-IDF Logistic Regression and calibrated Linear SVM. Group wording variants before splitting. Scores here measure performance on highly regular synthetic annotations, not real citizen behavior.

```python
from src.nlp.intent_classifier import train_and_evaluate
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

metrics = train_and_evaluate()
print("selected:", metrics["selected_model"])
print("validation:", metrics["validation"])
print("held-out test macro F1:", metrics["test"]["macro_f1"])
print("per language:", metrics["per_language_test"])
```

```python
labels = metrics["labels"]
cm = pd.DataFrame(metrics["test"]["confusion_matrix"], index=labels, columns=labels)
plt.figure(figsize=(13, 10))
sns.heatmap(cm, cmap="Blues", annot=True, fmt="d")
plt.xticks(rotation=75, ha="right")
plt.tight_layout()
plt.show()
class_rows = {label: metrics["test"]["per_class"][label] for label in labels}
pd.DataFrame(class_rows).T[["precision", "recall", "f1-score", "support"]]
```

## Reproducibility
Run `python -m src.nlp.train`. The script writes `models/nlp/intent_classifier.joblib` locally and JSON configuration/metrics. The artifact is ignored by Git.
