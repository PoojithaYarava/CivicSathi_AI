# CivicSathi AI: Project Setup
Phase 1 establishes schemas, reproducible synthetic examples, source provenance, and privacy controls. It does not implement the chatbot or NLP models.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
REQUIRED_DIRS = [
    'data/raw/public', 'data/raw/external', 'data/interim', 'data/processed',
    'documents/certificates', 'documents/municipal', 'documents/schemes',
    'documents/identity', 'documents/other_services', 'notebooks', 'src',
    'config', 'reports', 'tests',
]
for relative_path in REQUIRED_DIRS:
    (PROJECT_ROOT / relative_path).mkdir(parents=True, exist_ok=True)
sys.path.insert(0, str(PROJECT_ROOT))
print(f'Project root: {PROJECT_ROOT}')
```

```python
assert sys.version_info >= (3, 11), 'Python 3.11+ is required'
requirements = PROJECT_ROOT / 'requirements.txt'
assert requirements.exists(), 'Install requirements.txt before running data workflows'
print(f'Python {sys.version.split()[0]} is ready; dependencies are declared in {requirements.name}.')
```

## Privacy and reproducibility
No real citizen records, credentials, OTPs, exact home addresses, or private documents belong in this project. Raw inputs and interim extracts are Git-ignored. Synthetic examples must retain `source_type=synthetic_annotated`.

Create the starter CSV package with `python -m src.data_processing.build_datasets`, then run `pytest`. Source collection is an explicit, public-page-only action in notebook 2; review candidates before promoting them to processed data.
